```python
import pandas as pd
import joblib

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv(r"C:\Users\ARPIT DIMRI\OneDrive\Desktop\Premiere-League-pred-model\data\features.csv")
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

df.head()

```

```python
df.info()
```

```python
le_home = LabelEncoder()
le_away = LabelEncoder()
le_result = LabelEncoder()

df['home_encoded'] = le_home.fit_transform(df['Home'])
df['away_encoded'] = le_away.fit_transform(df['Away'])
df['result_encoded'] = le_result.fit_transform(df['result'])
```

```python
features = [
    'home_encoded',
    'away_encoded',
    'home_goals_avg',
    'away_goals_avg',
    'home_conceded_avg',
    'away_conceded_avg',
    'home_goal_diff',
    'away_goal_diff',
    'strength_diff',
    'attack_diff',
    'defense_diff'
]

X = df[features]
y = df['result_encoded']
```

```python
split_index = int(len(df) * 0.8)

X_train, X_test = X[:split_index],X[split_index:]
y_train, y_test = y[:split_index],y[split_index:]

print("Train Shape:", X_train.shape)
print("Test shape:",X_test.shape)
```

```python
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train,y_train)

lr_preds = lr_model.predict(X_test)
```

```python
print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_preds))
print(classification_report(y_test, lr_preds))
```

```python
#Model 2
rf_model = RandomForestClassifier(n_estimators=200, random_state=42)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
```

```python
print("Random Forest Accuracy:",accuracy_score(y_test, rf_preds))
print(classification_report(y_test, rf_preds))
```
